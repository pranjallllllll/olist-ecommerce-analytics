## **Step 4 : Investigate Missing Values**

The previous step showed that some columns contain missing values. Instead of
removing them immediately, I want to understand why they are missing and
whether the missing values are related to the order status.

I will compare the missing delivery dates across different order statuses to
identify which missing values are expected and which require further
investigation.

Most missing delivery dates are expected for orders that have not completed
the delivery process. However, some orders marked as `delivered` are missing
delivery dates, so these cases need to be investigated separately.

I will keep the records unchanged for now and decide how to handle them during

```python
# Group orders by their status
# Then count missing values in the two delivery date columns
missing_by_status = orders.groupby("order_status")[
    ["order_delivered_carrier_date", "order_delivered_customer_date"]
].apply(lambda x: x.isnull().sum())

# Display the result
missing_by_status
```

### **Observation**

Missing delivery dates are strongly related to the `order_status`, with many
missing values occurring in orders that have not completed the delivery
process.

However, some `delivered` orders are also missing delivery dates. These cases
need further investigation before deciding how they should be handled.

Therefore, missing delivery dates should not be removed or filled without
considering the order status.

## **Step 5 : Investigate the Unusual Missing Values**

A small number of orders are marked as `delivered` but are missing one or more
delivery dates.

I will inspect these orders to understand the missing information before
deciding how to handle them during data cleaning.

```python
# Check delivered orders with missing delivery dates
delivered_orders = orders[orders["order_status"] == "delivered"]

missing_customer_delivery = delivered_orders[
    delivered_orders["order_delivered_customer_date"].isnull()
]

missing_carrier_delivery = delivered_orders[
    delivered_orders["order_delivered_carrier_date"].isnull()
]

missing_both_delivery = delivered_orders[
    delivered_orders["order_delivered_customer_date"].isnull() &
    delivered_orders["order_delivered_carrier_date"].isnull()
]

print(
    "Delivered orders missing customer delivery date:",
    len(missing_customer_delivery)
)

print(
    "Delivered orders missing carrier delivery date:",
    len(missing_carrier_delivery)
)

print(
    "Delivered orders missing both delivery dates:",
    len(missing_both_delivery)
)
```

```python
# Display delivered orders missing the carrier delivery date
delivered_missing_carrier_date = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_carrier_date"].isnull())
]

display(delivered_missing_carrier_date)
```

### **Observation**

## **Step 18 : Check One-to-Many Order Relationships**

Several Olist datasets can contain multiple records for the same order.

sellers, payments, and reviews.

Understanding these relationships is important because directly joining
these tables can duplicate order-level records and inflate metrics such
as revenue.

## **Step 19 : Check Important Data-Quality Issues**

Before cleaning the data, I want to check for values that could affect the
accuracy of the analysis. These checks focus on order timelines, prices,
product weights, and review dates.

The goal is to identify potential data-quality issues now and decide how to
handle them during the data-cleaning stage.

```python
# Create a temporary copy so the original orders dataset is not modified
timeline = orders.copy()

# Convert order timestamp columns to datetime for timeline checks
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    timeline[col] = pd.to_datetime(timeline[col])

# Check for orders delivered before they were purchased
print(
    "Orders with delivery before purchase:",
    (
        timeline["order_delivered_customer_date"]
        < timeline["order_purchase_timestamp"]
    ).sum()
)

# Check for order items with zero or negative prices
print(
    "Order items with zero or negative prices:",
    (order_items["price"] <= 0).sum()
)

# Check for products with zero or unusually high weight
print(
    "Products with zero weight:",
    (products["product_weight_g"] == 0).sum()
)

print(
    "Products with weight above 30 kg:",
    (products["product_weight_g"] > 30000).sum()
)

# Compare review creation dates with the original order purchase dates
review_check = reviews.merge(
    orders[["order_id", "order_purchase_timestamp"]],
    on="order_id",
    how="left"
)

# Convert the dates to datetime for comparison
review_check["order_purchase_timestamp"] = pd.to_datetime(
    review_check["order_purchase_timestamp"]
)

review_check["review_creation_date"] = pd.to_datetime(
    review_check["review_creation_date"]
)

# Check for reviews created before the order was purchased
print(
    "Reviews created before purchase:",
    (
        review_check["review_creation_date"]
        < review_check["order_purchase_timestamp"]
    ).sum()
)
```

### **Observation**

No orders were found with a delivery date before the purchase date, and no
order items had a zero or negative price.

However, **4 products have a weight of 0 g**, which may represent missing or
incorrect product information. **1 product weighs more than 30 kg**, which may
be valid depending on the product and should not be treated as an error
automatically.

There are **74 reviews created before the order purchase date**. These records
will need to be investigated during the data-cleaning stage before using review
timing in the analysis.

No records are being removed at this stage. Potential issues will be handled
during data cleaning based on their context.
