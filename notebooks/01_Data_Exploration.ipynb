## **Step 4 : Investigate Missing Values**

The previous step showed that some columns contain missing values. Instead of
removing them immediately, I want to understand why they are missing and
whether the missing values are related to the order status.

I will first compare the missing delivery dates across different order
statuses. This will help me understand whether the missing values are
expected or unusual.

Most missing delivery dates appear to be expected for orders that have not
completed the delivery process. However, a small number of orders marked as
`delivered` are missing delivery dates, so I will investigate these cases
separately.

I will keep the records unchanged for now and decide how to handle them
during the data-cleaning stage.

```python
# Group orders by their status
# Then count missing values in the two delivery date columns
missing_by_status = orders.groupby("order_status")[
    ["order_delivered_carrier_date", "order_delivered_customer_date"]
].apply(lambda x: x.isnull().sum())

# Display the result
missing_by_status
```

### **Observations**

The missing delivery dates are largely related to the order status.

Orders that have not completed the delivery process, such as `shipped`,
`processing`, `canceled`, and `invoiced`, contain many missing delivery
dates. These missing values are generally expected because the corresponding
delivery stage may not have occurred.

However, a small number of orders marked as `delivered` are missing delivery
dates. These cases need further investigation before deciding how they should
be handled during data cleaning.

Therefore, I should not remove rows simply because they contain missing
delivery dates.

## **Step 5 : Investigate the Unusual Missing Values**

A small number of orders are marked as `delivered` but are missing one or more
delivery dates.

I will inspect these orders to understand the missing information before
deciding how to handle them during data cleaning.

I will keep these records unchanged for now. Orders with missing delivery
dates will only be excluded from calculations that specifically require those
dates.

```python
# Find delivered orders missing the customer delivery date
delivered_missing_customer_date = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_customer_date"].isnull())
]

# Display the affected orders
delivered_missing_customer_date
```

```python
# Find delivered orders missing the carrier delivery date
delivered_missing_carrier_date = orders[
    (orders["order_status"] == "delivered") &
    (orders["order_delivered_carrier_date"].isnull())
]

# Display the affected orders
delivered_missing_carrier_date
```

### **Observation**

The investigation found:

- **8 delivered orders** missing `order_delivered_customer_date`
- **2 delivered orders** missing `order_delivered_carrier_date`
- **1 order** is missing both dates

These orders will be kept in the dataset because the missing dates do not
make the entire order record unusable.
